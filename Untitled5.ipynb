```python
# Student Marks Analysis

students = ["Kalpana", "Priya", "Kavi", "Anu", "Divya"]
marks = [85, 92, 78, 88, 95]

# Calculate average
average = sum(marks) / len(marks)

# Find highest and lowest marks
highest = max(marks)
lowest = min(marks)

print("Student Marks")
print("----------------")

for i in range(len(students)):
    print(students[i], ":", marks[i])

print("\nAverage Mark:", average)
print("Highest Mark:", highest)
print("Lowest Mark:", lowest)
```

```python
# Sales Data Analysis

products = ["Laptop", "Mobile", "Tablet", "Headphone", "Keyboard"]

sales = [5, 12, 8, 20, 15]

prices = [50000, 20000, 15000, 2000, 1500]

print("PRODUCT SALES REPORT")
print("--------------------")

total_revenue = 0

for i in range(len(products)):
    revenue = sales[i] * prices[i]
    total_revenue += revenue

    print(products[i])
    print("Quantity:", sales[i])
    print("Price:", prices[i])
    print("Revenue:", revenue)
    print()

print("--------------------")
print("Total Revenue:", total_revenue)

highest_sales = max(sales)
index = sales.index(highest_sales)

print("Most Sold Product:", products[index])
print("Units Sold:", highest_sales)
```
